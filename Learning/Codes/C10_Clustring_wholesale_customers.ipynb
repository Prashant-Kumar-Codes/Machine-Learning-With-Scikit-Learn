## About Dataset

### Problem Statement

Customer Personality Analysis is a detailed analysis of a company’s ideal customers. It helps a business to better understand its customers and makes it easier for them to modify products according to the specific needs, behaviors and concerns of different types of customers.

Customer personality analysis helps a business to modify its product based on its target customers from different types of customer segments. For example, instead of spending money to market a new product to every customer in the company’s database, a company can analyze which customer segment is most likely to buy the product and then market the product only on that particular segment.

### Feature Engineering

First creating some features out of `Dt_Customer` which is the date of enrollment of customer with the company

```python
print(f'Type of Dt_Customer: {data["Dt_Customer"].dtype}\n')      ## Categorical Column of 
print(f'Dt_Customer before changing to DataTime: {data['Dt_Customer'].head()}')
```

```python

data['Dt_Customer'] = pd.to_datetime(data['Dt_Customer'], format='mixed')
print(f'Dt_Customer after changing to DataTime:\n{data['Dt_Customer'].head()}')

print(f'\nOldest date: {data.Dt_Customer.min()}')
print(f'Newest date: {data.Dt_Customer.max()}')
```

Creating a Customer_Days column which contains the number of days the customer has been in the company w.r.t to the latest joined customer

```python
dates = data['Dt_Customer']
print(f'{dates}')

d1 = max(dates)
print(f'\n\nd1: {d1}')

days = (d1 - dates).dt.days

data['Customer_Days'] = days
data['Customer_Days'] = pd.to_numeric(data['Customer_Days'], errors='coerce')

print(f'\n\nCustomer_Days column:\n{data['Customer_Days']}')
```
