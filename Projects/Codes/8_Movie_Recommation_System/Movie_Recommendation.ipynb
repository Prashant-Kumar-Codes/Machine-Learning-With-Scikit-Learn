### 2. Data to Required Format

#### a. genres

```python
df.iloc[0].genres
```

```python
def genres_dic_to_list(row):
    genre_name = []
    for genre in ast.literal_eval(row):
        genre_name.append(genre.get('name', '<unk>'))
    return genre_name

df['genres'] = df['genres'].apply(genres_dic_to_list)
```

```python
df['genres'].head()
```

#### b. keywords

```python
df.iloc[0].keywords
```

```python
def movies_key_to_list(row):
    keywords = []
    for keyword in ast.literal_eval(row):
        keywords.append(keyword.get('name', '<unk>'))
    return keywords

df['keywords'] = df['keywords'].apply(movies_key_to_list)
```

#### d. crew

```python
def movies_crew_extract_director(row):
    crews = []
    for crew in ast.literal_eval(row):
        if crew['job'] == 'Director':
            crews.append(crew.get('name', '<unk>'))
            return crews
    
df['crew'] = df['crew'].apply(movies_crew_extract_director)
```

#### **Removing Spaces between the words to remove the complexity due to space**

```python
df['genres'] = df['genres'].apply(lambda x: [i.replace(' ', '') for i in x])
df['keywords'] = df['keywords'].apply(lambda x: [i.replace(' ', '') for i in x])
df['cast'] = df['cast'].apply(lambda x: [i.replace(' ', '') for i in x])
df['crew'] = df['crew'].apply(lambda x: [i.replace(' ', '') for i in x])
df['overview'] = df['overview'].apply(lambda x: [i.replace(' ', '') for i in x])
```
